```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv('/content/17customer_reviews.csv', encoding='latin1')

review = df['Review Text'].str.lower().str.split().explode().value_counts()
remove=["the","and","is"]
rem_w= review[~review.index.isin(remove)]
print(rem_w)
print(rem_w.nlargest(5))

review=rem_w.nlargest(5)
plt.bar(review.index,review.values)
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.title("Most Frequent Words")
plt.show()
```

```python
import pandas as pd
import numpy as np

df=pd.read_csv('/content/19_blood_pressure_trial_500_patients - 19_blood_pressure_trial_500_patients.csv')

df["Group"] = df["Group"].str.strip()
df["BP_Difference"] = df["Before_BP_mmHg"] - df["After_BP_mmHg"]

drugA = df[df["Group"] == "Drug"]["BP_Difference"]
placebo = df[df["Group"] == "Placebo"]["BP_Difference"]

mean_drug = drugA.mean()
mean_placebo = placebo.mean()

std_drug = drugA.std()
std_placebo = placebo.std()

n_drug = drugA.count()
n_placebo = placebo.count()

se_drug = std_drug / np.sqrt(n_drug)
se_placebo = std_placebo / np.sqrt(n_placebo)

z = 1.96

me_drug = z * se_drug
me_placebo = z * se_placebo

ci_drug = (mean_drug - me_drug, mean_drug + me_drug)
ci_placebo = (mean_placebo - me_placebo, mean_placebo + me_placebo)

print("Drug Group")
print("Mean BP Difference:", round(mean_drug,2))
print("95% Confidence Interval:", (round(ci_drug[0],2), round(ci_drug[1],2)))

print("\nPlacebo Group")
print("Mean BP Difference:", round(mean_placebo,2))
print("95% Confidence Interval:", (round(ci_placebo[0],2), round(ci_placebo[1],2)))
```

```python

import pandas as pd
import numpy as np

df=pd.read_csv('/content/20_AB_Testing_Website_Dataset - 20_AB_Testing_Website_Dataset.csv')

A = df[df['Design']=='A']['Converted']
B = df[df['Design']=='B']['Converted']
p1, p2 = A.mean(), B.mean()
p = (A.sum()+B.sum())/(len(A)+len(B))
z = (p2-p1)/np.sqrt(p*(1-p)*(1/len(A)+1/len(B)))
print("Statistically significant" if abs(z)>1.96 else "Not statistically significant")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv('/content/18bodyfat FINAL - 18bodyfat FINAL.csv')

print("The mean of Age is:", round(df['Age'].mean(), 2))
print("The mean of Bodyfat is:", round(df['BodyFat'].mean(), 2))
print("The median Age is:", df['Age'].median())
print("The median Bodyfat is:", df['BodyFat'].median())
print("The Standard Deviation Age is:", round(df['Age'].std(), 2))
print("The Standard Deviation Bodyfat is:", round(df['BodyFat'].std(), 2))

plt.boxplot([df['Age'], df['BodyFat']], labels=['Age', 'Body Fat %'])
plt.title("Distribution of Age and Body Fat")
plt.show()

plt.scatter(df['Age'], df['BodyFat'],marker="o")
plt.title("Age vs Body Fat")
plt.xlabel("Age")
plt.ylabel("Body Fat %")
plt.show()
```
